```python
import numpy as np 
class HebbNetwork:
    def __init__(self, input_size):
        self.weights = np.zeros(input_size)
        self.bias = 0
    def train(self,X,Y):
        print("Tranging Process:\n")
        for x, y in zip(X,Y):
            self.weights += x*y
            self.bias += y
            print("Input : ", x)
            print("Target : ", y)
            print("Updated Weights: ", self.weights)
            print("Updated Bias : ", self.bias)
            print("_"*40)
    def predict(self,x):
        activation = np.dot(self.weights, x) + self.bias
        if activation >= 0:
            return 1 
        else:
            return -1
X = np.array([[-1,-1],[-1,1],[1,-1],[1,1]])
Y = np.array([-1,1,1,1])
network = HebbNetwork(input_size=2)
network.train(X,Y)
print("\nFinal Weights: ", network.weights)
print("Final Bais: ", network.bias)
print("\nTesting OR Gate using Hebb Rule\n")

for x, y in zip(X,Y):
    prediction = network.predict(x)
    print("Input:",x,"Predicted:", prediction,"Actual:",y)
    
            
```
